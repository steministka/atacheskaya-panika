# Контроль конфаундеров, анализ специфичности на группах «нормальное старение» и «соматоформные расстройства»

### Назначение: 
Контроль конфаундеров и анализ специфичности.
### Содержимое:
- Попытка предсказать возраст по тем же признакам внутри контрольной группы.
- Метрики модели отдельно по возрастным подгруппам.
- Проверка, что оценка модели не сводится к одному поведенческому показателю (например, суммарному времени решения таблиц Шульте).
- Анализ специфичности: проверка, что модель даёт низкую вероятность ПТСР на группах «нормальное старение» (65+) и «соматоформные расстройства».
### Обязательно: 
подробное описание и доказательная база всех представленных выводов.
